```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain_community.tools import DuckDuckGoSearchRun
from langchain_core.messages import HumanMessage
import requests
from dotenv import load_dotenv
load_dotenv()
```

```python
@tool
def multiply(a: int, b: int) -> int:
    """Given two inputs this tool returns the prodcut of the inputs"""
    return a*b
```

```python
print(multiply.invoke({'a':3, 'b': 4}))
```

```python
llm = ChatOpenAI()
```

```python
llm_binded = llm.bind_tools([multiply])
```

```python
llm_binded.invoke('how are you')
```

```python
result = llm_binded.invoke('multiply 2 with 20').tool_calls
```
